```python
# ==========================================
# Name: Faiza Memon
# TASK 1: EXPLORING AND VISUALIZING IRIS DATASET
# Dataset: Iris Dataset
# DevelopersHub Data Science Internship

# ==========================================
# -----------------------------
# IMPORT LIBRARIES
# -----------------------------
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# -----------------------------
# LOAD DATASET
# -----------------------------
iris = sns.load_dataset('iris')

# -----------------------------
# DATASET UNDERSTANDING
# -----------------------------
print("="*50)
print("DATASET SHAPE")
print("="*50)
print(iris.shape)

print("\n" + "="*50)
print("COLUMN NAMES")
print("="*50)
print(iris.columns)

print("\n" + "="*50)
print("FIRST 5 ROWS")
print("="*50)
print(iris.head())

print("\n" + "="*50)
print("DATASET INFORMATION")
print("="*50)
print(iris.info())

print("\n" + "="*50)
print("STATISTICAL SUMMARY")
print("="*50)
print(iris.describe())

# -----------------------------
# DATA CLEANING
# -----------------------------
print("\n" + "="*50)
print("MISSING VALUES")
print("="*50)
print(iris.isnull().sum())

print("\n" + "="*50)
print("DUPLICATE ROWS")
print("="*50)
print(iris.duplicated().sum())

# Remove duplicates if any
iris = iris.drop_duplicates()

# -----------------------------
# SCATTER PLOT
# -----------------------------
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()

# -----------------------------
# HISTOGRAM
# -----------------------------
plt.figure(figsize=(8,6))

sns.histplot(
    iris['sepal_length'],
    bins=10,
    kde=True
)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()

# -----------------------------
# BOX PLOT
# -----------------------------
plt.figure(figsize=(8,6))

sns.boxplot(
    data=iris,
    x='species',
    y='sepal_length'
)

plt.title("Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.show()

# -----------------------------
# ADDITIONAL VISUALIZATION
# -----------------------------
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=iris,
    x='petal_width',
    y='petal_length',
    hue='species'
)

plt.title("Petal Width vs Petal Length")
plt.show()

# -----------------------------
# CORRELATION HEATMAP
# -----------------------------
plt.figure(figsize=(8,6))

numeric_data = iris.drop('species', axis=1)

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap='Blues'
)

plt.title("Correlation Heatmap")
plt.show()

# -----------------------------
# CONCLUSION
# -----------------------------
print("\n" + "="*50)
print("CONCLUSION")
print("="*50)

print("""
1. Iris dataset contains 150 observations.
2. The dataset has four numerical features and one target column.
3. No missing values were found.
4. Scatter plots show relationships between flower measurements.
5. Histograms reveal feature distributions.
6. Box plots help identify spread and potential outliers.
7. Petal measurements are strongly correlated.
""")
```
